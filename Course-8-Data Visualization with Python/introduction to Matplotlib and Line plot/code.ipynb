```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_excel(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Canada.xlsx',
    sheet_name='Canada by Citizenship',
    skiprows=range(20),
    skipfooter=2)

print('Data read into a pandas dataframe!')
```

```python
df.drop(['Type', 'Coverage','AREA','REG','DEV'], axis=1, inplace=True)
df
```

```python
df.rename(columns={'OdName':'Country', 'AreaName':'Continent', 'RegName':'Region'}, inplace=True)
df.columns
```

```python
numeric_df = df.select_dtypes(include='number')
df['Total'] = numeric_df.sum(axis=1)
df
```

```python
df.head()
```

```python
df.set_index("Country", inplace=True)
```

```python
df.index.name = None
df
```

```python
df.loc["Japan"]
```

```python
df.columns = list(map(str, df.columns))
```

```python
df.columns
```

```python
year = list(map(str, range(1990, 2014)))
haiti = df.loc['Haiti', year]
haiti
```

```python
df.sort_values(by="Total", ascending=False, inplace=True)
df.head()
```

```python
df.sort_values(by="2010", ascending=False, inplace=True)
df.head()
```

```python
%matplotlib inline 

import matplotlib.pyplot as plt
```

```python
#Since we converted the years to string, 
#let's declare a variable that will allow us to easily call upon the full range of years:
years = list(map(str, range(1980, 2014)))
#creating data series
haiti = df.loc['Haiti', years] 
haiti.plot( kind="line")
plt.title('Immigration from Haiti')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')

plt.show()
```

```python
haiti.index = haiti.index.map(int) 
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# annotate the 2010 Earthquake. 
# syntax: plt.text(x, y, label)
plt.text(2000, 6000, '2010 Earthquake') # see note below

plt.show()
```
