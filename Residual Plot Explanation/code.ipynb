### Import important libraries

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

### What is Residual?
Residual = Actual value - Predicted value

<h1 style="text-align: center; color: white; font-weight: bold; background: grey">Case 1: Good Fit Model</h1>

### Residuals will be around 0 line

```python
# Linear data
X = np.array([10, 20, 30, 40, 50]).reshape(-1, 1)
y = np.array([15, 25, 35, 45, 55])

# Model
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

# Residuals
residuals = y - y_pred

sns.residplot( x= y_pred, y=residuals)
plt.title("Residual Plot: Good Fit")
plt.xlabel("y_pred")
plt.ylabel("Residuals")
plt.show()
```

<h1 style="text-align: center; color: white; font-weight: bold; background: grey">Case 2: Non-Linear Relationship</h1>

### Residuals will create a curve or U-shape

```python
# Non-linear data
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([1, 4, 9, 16, 25, 36])  # y = x^2

# Linear model (wrong choice)
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

# Residuals
residuals = y - y_pred


sns.residplot(x=y_pred, y=residuals)
plt.title("Residual Plot: Non-Linear Relationship")
plt.xlabel("y_pred")
plt.ylabel("Residuals")
plt.show()
```

<h1 style="text-align: center; color: white; font-weight: bold; background: grey">Case 3: Unequal Variance (Heteroscedasticity)</h1>

### Residuals create Funnel Shape

```python
# Data with increasing noise
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([2, 4, 8, 16, 32, 64]) + np.array([0, 1, 3, 6, 10, 15])

# Model
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

# Residuals
residuals = y - y_pred

sns.residplot(x=y_pred, y=residuals)
plt.title("Residual Plot: Unequal Variance (Heteroscedasticity)")
plt.xlabel("y_pred")
plt.ylabel("Residuals")
plt.show()
```
